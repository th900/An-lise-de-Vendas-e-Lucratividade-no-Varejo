```python
import pandas as pd
import numpy as np

df = pd.read_csv('/kaggle/input/superstore-dataset-final/Sample - Superstore.csv', encoding='latin1')
```

```python
df = df.drop_duplicates()
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Ship Date"] = pd.to_datetime(df["Ship Date"])
```

```python
df.info()
```

```python
print("Período dos pedidos:")
print(df["Order Date"].min(), "até", df["Order Date"].max())

print("\nPeríodo dos envios:")
print(df["Ship Date"].min(), "até", df["Ship Date"].max())

print("\nDatas de envio anteriores ao pedido:")
print((df["Ship Date"] < df["Order Date"]).sum())
```

```python
df["Shipping days"] = (df["Ship Date"] - df["Order Date"]).dt.days
df["Order Year"] = df["Order Date"].dt.year
df["Order Month"] = df["Order Date"].dt.month
df["Order Month Name"] = df["Order Date"].dt.month_name()
```

```python
df["profit margin"] = df["Profit"]/df["Sales"]
```

```python
df["Discount"].describe()
```

```python
df.loc[df["Discount"] > 0.70, "profit margin"]
```

```python
discount_analysis = (
    df.groupby("Discount")
      .agg(
          Sales=("Sales", "sum"),
          Profit=("Profit", "sum"),
          Orders=("Order ID", "nunique")
      )
)

discount_analysis["profit margin"] = (
    discount_analysis["Profit"] /
    discount_analysis["Sales"]
)

discount_analysis
```

```python
df[df["Discount"] > 0.70].agg({
    "Sales": "sum",
    "Quantity": "sum",
    "Profit": "sum"
})
```

```python
high_discount = (
    df[df["Discount"] >= 0.30]
    .groupby(["Category", "Sub-Category"])
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Quantity=("Quantity", "sum"),
        Orders=("Order ID", "nunique")
    )
)

high_discount["profit margin"] = (
    high_discount["Profit"] /
    high_discount["Sales"]
)

high_discount.sort_values("profit margin")
```

```python
subcategory_discount = (
    df.groupby(["Sub-Category", "Discount"])
      .agg(
          Sales=("Sales", "sum"),
          Profit=("Profit", "sum"),
          Orders=("Order ID", "nunique")
      )
)

subcategory_discount["Profit Margin"] = (
    subcategory_discount["Profit"] /
    subcategory_discount["Sales"]
)

subcategory_discount
```

```python
regional_analysis = (
    df.groupby("Region")
      .agg(
          Sales=("Sales", "sum"),
          Profit=("Profit", "sum"),
          Quantity=("Quantity", "sum"),
          Orders=("Order ID", "nunique"),
          Customers=("Customer ID", "nunique")
      )
)

regional_analysis["Profit Margin"] = (
    regional_analysis["Profit"] /
    regional_analysis["Sales"]
)
regional_analysis.sort_values("Profit", ascending=False)
```

```python
region_discount = (
    df.groupby(["Region", "Discount"])
      .agg(
          Sales=("Sales", "sum"),
          Profit=("Profit", "sum"),
          Orders=("Order ID", "nunique")
      )
)

region_discount["Profit Margin"] = (
    region_discount["Profit"] /
    region_discount["Sales"]
)

region_discount
```

```python
region_category = (
    df.groupby(["Region", "Category"])
      .agg(
          Sales=("Sales", "sum"),
          Profit=("Profit", "sum"),
          Orders=("Order ID", "nunique")
      )
)

region_category["Profit Margin"] = (
    region_category["Profit"] /
    region_category["Sales"]
)

region_category
```

Central : Furniture tem margem negativa e Office Supplies está bem abaixo quando comparamos com as outras regiões. Isso pode estar relacionado com desconto?

```python
central_furniture = (
    df[
        (df["Region"] == "Central") &
        (df["Category"] == "Furniture")
    ]
    .groupby("Discount")
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Orders=("Order ID", "nunique")
    )
)

central_furniture["Profit Margin"] = (
    central_furniture["Profit"] /
    central_furniture["Sales"]
)

central_furniture
```

```python
central_office = (
    df[
        (df["Region"] == "Central") &
        (df["Category"] == "Office Supplies")
    ]
    .groupby("Discount")
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Orders=("Order ID", "nunique")
    )
)

central_office["Profit Margin"] = (
    central_office["Profit"] /
    central_office["Sales"]
)

central_office
```

```python
central_furniture_sub = (
    df[
        (df["Region"] == "Central") &
        (df["Category"] == "Furniture")
    ]
    .groupby(["Sub-Category", "Discount"])
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Orders=("Order ID", "nunique")
    )
)

central_furniture_sub["Profit Margin"] = (
    central_furniture_sub["Profit"] /
    central_furniture_sub["Sales"]
)

central_furniture_sub.sort_values("Profit")
```

```python
central_office_80 = (
    df[
        (df["Region"] == "Central") &
        (df["Category"] == "Office Supplies") &
        (df["Discount"] == 0.80)
    ]
    .groupby("Sub-Category")
    .agg(
        Sales=("Sales", "sum"),
        Profit=("Profit", "sum"),
        Orders=("Order ID", "nunique")
    )
)

central_office_80["Profit Margin"] = (
    central_office_80["Profit"] /
    central_office_80["Sales"]
)

central_office_80.sort_values("Profit")
```

1. Desconto × rentabilidade

Quanto maior o desconto, menor tende a ser a margem, com vários níveis tornando-se deficitários.

2. Limite de desconto varia por subcategoria

Não existe um único desconto "seguro" para todos os produtos.

3. Furniture no Central

A categoria é lucrativa sem desconto, mas passa a gerar prejuízo em níveis elevados de desconto.

4. Chairs

Um dos impactos mais relevantes dentro de Furniture: 95 pedidos a 30% de desconto geraram prejuízo.

5. Central + Office Supplies + 80%

250 pedidos geraram -$30.539 de prejuízo.

6. Binders

É o principal responsável pelo prejuízo desse grupo, com -$21.909.
7. Appliances

Tem a pior margem nesse grupo (-255,1%), mas seu impacto absoluto é menor que o de Binders.

8. Região

Central possui a menor margem regional (7,92%), enquanto West possui a maior (14,94%).

9. Categoria × região

O problema do Central não está distribuído igualmente entre categorias; Furniture é negativa e Office Supplies apresenta margem bem inferior às demais regiões.

10. Volume ≠ rentabilidade

Uma região ou categoria pode ter vendas maiores e ainda assim gerar menos lucro.
